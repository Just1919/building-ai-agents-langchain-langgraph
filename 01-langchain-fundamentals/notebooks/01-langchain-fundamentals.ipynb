## 1. What Are LLM Applications?

In recent years, Large Language Models (LLMs) such as ChatGPT, Gemini, and Claude have transformed the way we interact with computers. Instead of writing complex rules and conditions, we can simply ask questions in natural language and receive useful answers.

An **LLM application** is any piece of software that uses such a model to:

- Understand and generate human-like text.
- Automate reasoning tasks.
- Act as a digital assistant for a specific purpose.

### Real-World Examples

- A chatbot that answers questions about your company's documents.
- An AI tutor that explains coding concepts.
- An assistant that summarizes emails or generates weekly reports.
- A support agent that classifies tickets and drafts replies.

The key insight is this:

**An LLM application is never just a model.**

It is a combination of:

**AI Model + Logic + Data + Tools**

Think of the model as the engine, but the application as the entire car.

## 2. Why We Need Frameworks Like LangChain and LangGraph

For a small prototype, calling a model API directly is perfectly fine. However, as soon as your application requires any of the following, things can quickly become complicated:

- **Multiple steps:** Summarize → Analyze → Generate a report.
- **Memory:** Remembering what was said earlier.
- **Tool use:** Searching the web or querying a database.
- **Error handling and retries:** Managing failures and recovering from unexpected errors.
- **Provider flexibility:** Switching from one model provider to another.
- **Observability:** Understanding what happened when something goes wrong.

This is where **LangChain** and **LangGraph** come in.

### 2.1. LangChain: The Toolbox

LangChain helps you build structured workflows around LLMs using reusable components such as:

- Prompts
- Models
- Output parsers
- Chains
- Tools
- Agents

Think of LangChain as a **toolbox for building AI-powered applications and pipelines**.

### 2.2. LangGraph: The Control Flow

LangGraph builds on the LangChain ecosystem. It allows you to design **graph-based workflows** where each step (a *node*) processes information, while *edges* determine what happens next.

It is particularly useful for:

- **Multi-agent systems:** Researcher → Writer → Editor.
- **Stateful applications:** Maintaining information throughout a workflow.
- **Error-tolerant workflows:** Implementing loops, conditional logic, and recovery mechanisms.

Imagine a flowchart where each box represents a function or an AI agent, and the arrows determine how information moves between them.

> **Mental Model**
>
> - **LangChain** provides the components.

## 3. The Four Key Components of an LLM Application

Every LLM application, from a simple Python script to a production-ready AI agent, is built around four fundamental ingredients.

### 3.1. Prompt

The **prompt** is the input text or instruction provided to the model.

*Example:*

"Summarize this article in three sentences."

### 3.2. Model

The **model** is the LLM responsible for processing the input and generating a response.

*Examples:* GPT-4o, Gemini, Claude.

### 3.3. Memory

**Memory** allows an application to retain information from previous interactions and use it in subsequent steps.

*Example:* Remembering a user's name throughout a conversation.

### 3.4. Tool / Function

A **tool** or **function** allows an LLM application to interact with external systems and perform actions.

*Examples:*

- Searching the web.
- Retrieving information from a database.
- Calling an external API.
- Executing a Python function.

LangChain and LangGraph help you assemble these components into structured applications without having to rewrite the same boilerplate code every time.




## 4. A Simple Example Without LangChain

Before introducing LangChain, let's see what a direct API call looks like using the **OpenAI Python SDK**.

This example sends a message directly to an OpenAI model and prints the generated response.

````python
# Import the OpenAI client
from openai import OpenAI



# Initialize the client
# The client automatically reads OPENAI_API_KEY from the environment
client = OpenAI()

# Send a request to the chat completion API
response = client.chat.completions.create(
    # Specify the language model to use
    model="gpt-4o-mini",

    # Provide the conversation messages
    messages=[
        {
            "role": "user",
            "content": "Explain LangChain in simple terms"
        }
    ],
)

# Extract and print the generated response
print(response.choices[0].message.content)
````


This works perfectly. However, imagine that you now want to:

- Add a **summarization step** before generating the answer.
- **Store the chat history** to maintain conversational context.
- Call a **web API** to fetch fresh data.
- **Switch to another model provider** next month.

Each of these requirements introduces additional code, and that code can quickly become difficult to manage.

Prompts get duplicated, parsing logic becomes scattered across the application, and switching providers may require rewriting several parts of your code.

This is exactly the problem **LangChain** was built to solve: keeping LLM applications **organized, modular, and scalable**.

## 7. Your First LangChain Example: A Text Summarizer

Let's build a simple text summarizer step by step.

### Step 1: Import the Components


````python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()
````


`langchain_openai` provides the integration with OpenAI models, while `langchain_core` contains the framework's fundamental abstractions: prompts, parsers, messages, and runnables.

### Step 2: Create the Model


````python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
````


The `temperature` parameter controls the creativity of the model's responses. A value of `0` makes the output as deterministic as possible, which is ideal for tasks such as extraction, classification, and summarization. Higher values (0.7–1.0) are better suited for creative writing.

Let's call the model directly to see what a chat model returns:


````python
response = llm.invoke("Explain what an LLM is in one sentence.")

print(type(response))        # <class 'langchain_core.messages.ai.AIMessage'>
print(response.content)      # the generated text
print(response.usage_metadata)  # token usage: input, output, total
````

Notice that the result is an **`AIMessage`** object, not a plain string. It contains the generated text (`.content`) as well as metadata such as token usage, which is useful for monitoring API costs.

### Step 3: Create a Prompt Template

A prompt template defines *what you want the model to do*, using placeholders that you can fill in later.

````python
prompt = ChatPromptTemplate.from_template(
    "Summarize this text in one sentence:\n\n{text}"
)
````

You can inspect what the template produces before sending it to any model:

````python
messages = prompt.invoke({
    "text": "LangChain is a powerful framework designed to simplify the development of applications powered by large language models. It provides developers with reusable components for building complex workflows, integrating external tools, connecting to data sources, and managing interactions with AI models. By combining these capabilities, developers can create intelligent applications such as chatbots, document assistants, and autonomous AI agents more efficiently."
})

print(messages)
````

This is a useful debugging habit: **always inspect the prompt before blaming the model.** Checking the prompt helps you verify that the instructions and input variables are correctly formatted before sending them to the LLM.

### Step 4: Combine Everything into a Chain

LangChain uses the pipe operator (`|`) to connect different components into a single workflow. This syntax is called the **LangChain Expression Language (LCEL)**, and it allows you to build modular, reusable chains by combining prompts, language models, and output parsers.

````python
chain = prompt | llm | StrOutputParser()
````

### What Happens When the Chain Runs?

When the chain is executed, LangChain processes the data through each component in sequence:

1. **`prompt`** fills in the template with the provided input variables and formats the messages for the model.
2. **`llm`** sends these messages to the language model and returns an `AIMessage` object containing the generated response and associated metadata.
3. **`StrOutputParser()`** extracts the text content from the `AIMessage` and converts it into a plain Python string.

### Step 5: Run the Chain

````python
summary = chain.invoke({
    "text": "LangChain helps developers build AI applications faster by "
            "simplifying prompt management and chaining logic."
})

print(summary)
````

That is your first working LangChain pipeline.

### 8. How It Works Internally

Behind the single line of code `prompt | llm | StrOutputParser()`, several operations take place in sequence. Each component processes the data and passes its output to the next component.

## 9. Going Further with Prompt Templates

In real-world applications, prompts often need to do more than provide a simple instruction to a language model. They may need to define the model's role, specify behavioral constraints, and handle different types of user input.

LangChain provides flexible prompt templates that make it easier to structure, customize, and reuse these instructions.

Let's explore three useful patterns.

### Pattern 1: System + Human Messages

A **system message** defines the model's role, behavior, and constraints, while a **human message** carries the actual user input.

Separating these two types of messages helps organize instructions and gives you greater control over how the model responds.

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a concise technical assistant. Answer in at most {max_sentences} sentences."),
    ("human", "{question}"),
])

chain = prompt | llm | StrOutputParser()

print(chain.invoke({
    "max_sentences": 2,
    "question": "What is a vector database?",
}))
````

Separating the *behavior* (system message) from the *input* (human message) is a recommended best practice when designing prompts. It makes your prompts easier to maintain, customize, and reuse across different applications.

### Pattern 2: Multiple Variables, One Template

Prompt templates can accept multiple input variables, allowing you to dynamically customize their content without changing the overall structure.

This means you can define a single reusable template and use it for different tasks by simply providing different values for each variable.

````python
translator_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a professional translator. Keep the tone {tone}."),
    ("human", "Translate the following text from {source_lang} to {target_lang}:\n\n{text}"),
])

translator = translator_prompt | llm | StrOutputParser()

print(translator.invoke({
    "tone": "formal",
    "source_lang": "English",
    "target_lang": "French",
    "text": "Our new feature will be available next week.",
}))
````

### Pattern 3: Question Answering

The same approach can also be used to build question-answering applications. You can define a reusable prompt template that instructs the model to answer questions based on a given context.

This is particularly useful for applications such as document assistants, knowledge-based chatbots, and Retrieval-Augmented Generation (RAG) systems.

````python
qa_prompt = ChatPromptTemplate.from_template(
    "Answer the question briefly.\n\nQuestion: {question}"
)

qa_chain = qa_prompt | llm | StrOutputParser()

print(qa_chain.invoke({"question": "What is LangChain?"}))
````

The lesson: the pipeline stays the same, only the template changes. That is the power of reusable prompts.

## 10. Structured Output: From Text to Data

Free-form text is useful for human-readable responses, but real-world applications often need **structured data** that Python can process directly.

Instead of manually extracting information from generated text, LangChain allows you to define an expected output structure and use the model's structured output capabilities to return data in a predictable format.

This makes it easier to validate, access, and integrate model responses into your Python applications.

````python
from typing import Literal
from pydantic import BaseModel, Field


class TicketAnalysis(BaseModel):
    """Analysis of a customer support ticket."""

    category: Literal["billing", "technical", "account", "other"] = Field(
        description="The main category of the ticket"
    )
    urgency: Literal["low", "medium", "high"] = Field(
        description="How urgent the ticket is"
    )
    summary: str = Field(description="A one-sentence summary of the problem")


structured_llm = llm.with_structured_output(TicketAnalysis)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You analyze customer support tickets."),
    ("human", "{ticket}"),
])

ticket_chain = prompt | structured_llm

result = ticket_chain.invoke({
    "ticket": "I was charged twice this month and I need a refund before Friday!"
})

print(result)
print(result.category, result.urgency)
````

The result is a real TicketAnalysis object, so you can write result.urgency == "high" in your code without any string parsing. This pattern is one of the most useful tools you will use when building agents, as we will see in the next articles.

## 11. Why Chains Are Powerful: Composing Multiple Steps

One of the main advantages of LangChain is the ability to combine multiple processing steps into a single workflow. Instead of asking the model to perform everything in one prompt, you can break a complex task into smaller, independent steps and connect them together.

For example, consider a pipeline that performs the following operations:

1. **Step 1 – Summarization:** Summarize the original text.
2. **Step 2 – Keyword Extraction:** Identify the main keywords from the generated summary.
3. **Step 3 – Title Generation:** Generate a concise and relevant title based on the summary.

By connecting these steps, you can build a modular pipeline where the output of one step becomes the input of the next.

### A Common Pitfall

A naive approach is to pipe two chains together directly. However, this can lead to unexpected results if the output of the first chain does not match the input expected by the second chain.

````python
# ⚠️ This does NOT work as expected
final_chain = summary_chain | keyword_chain
````

Why does it fail? summary_chain outputs a string, but keyword_chain starts with a prompt template that expects a dictionary such as {"summary": "..."}. The data shape does not match.

The right way: map the output to the next input

````python
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize this text in two sentences:\n\n{text}"
)
keyword_prompt = ChatPromptTemplate.from_template(
    "Extract 5 keywords from this summary, as a comma-separated list:\n\n{summary}"
)

parser = StrOutputParser()

summary_chain = summary_prompt | llm | parser

# Wrap the first chain in a dict so its output becomes the "summary" variable
keyword_chain = {"summary": summary_chain} | keyword_prompt | llm | parser

print(keyword_chain.invoke({
    "text": "LangChain is an open-source framework for building LLM applications. "
            "It provides standard interfaces for models, prompts, and tools, "
            "and integrates with hundreds of providers."
}))
````

In LCEL, a dictionary of runnables is automatically converted into a **`RunnableParallel`**. This means that each runnable is executed independently, and its output is collected into a single Python dictionary.

This is particularly useful when different steps require the same input but produce different outputs. It allows you to combine their results and pass them to subsequent steps in your pipeline.

### Running Steps in Parallel

When two steps are **independent** of each other, you can execute them simultaneously rather than waiting for one to finish before starting the other.

This can improve execution efficiency, especially when working with multiple LLM calls or independent processing tasks.

````python
from langchain_core.runnables import RunnableParallel

keywords_prompt = ChatPromptTemplate.from_template(
    "Extract 5 keywords from this text, comma-separated:\n\n{text}"
)
sentiment_prompt = ChatPromptTemplate.from_template(
    "Is the tone of this text positive, neutral, or negative? Answer with one word:\n\n{text}"
)

analysis = RunnableParallel(
    summary=summary_chain,
    keywords=keywords_prompt | llm | parser,
    sentiment=sentiment_prompt | llm | parser,
)

result = analysis.invoke({"text": "LangChain makes building LLM apps a joy. The docs are great!"})
print(result)
# {'summary': '...', 'keywords': '...', 'sentiment': 'Positive'}
````

All three branches receive the same input and execute concurrently. This reduces the overall execution time compared to running each step sequentially, since the independent operations do not need to wait for one another.

### Adding Your Own Python Logic

Not every step in a LangChain pipeline requires a language model. You can also integrate your own Python functions into a chain using **`RunnableLambda`**.

This allows you to combine traditional Python logic with LLM-powered components, making your pipelines more flexible and suitable for a wider range of tasks.

````python
from langchain_core.runnables import RunnableLambda

def to_keyword_list(raw: str) -> list[str]:
    return [k.strip().lower() for k in raw.split(",") if k.strip()]

keyword_list_chain = (
    keywords_prompt | llm | parser | RunnableLambda(to_keyword_list)
)

print(keyword_list_chain.invoke({"text": "LangChain simplifies building LLM apps."}))
# ['langchain', 'llm', 'apps', 'framework', 'python']
````

This is how AI steps and deterministic code live side by side in the same pipeline.

## 12. Understanding Runnables

You have already been working with Runnables since your very first LangChain example, even if you haven't explicitly encountered the concept by name.

In LangChain, **every component follows a common interface called `Runnable`**. This includes prompt templates, language models, output parsers, chains, and even custom Python functions wrapped with `RunnableLambda`.

This is one of the most important abstractions in the framework. Because all these components share the same interface, **they can be easily connected and combined**, allowing you to build complex workflows from smaller, reusable building blocks. In fact, every chain you create is itself a runnable.

Every runnable provides a consistent set of core methods that allow you to execute operations, process multiple inputs, stream results, and inspect intermediate outputs:

````python
chain.invoke(input)      # run on one input
chain.batch(inputs)      # run on a list of inputs
chain.stream(input)      # stream the output as it is generated
await chain.ainvoke(input)  # async version of invoke
````

#### Creating a Chain

Let's create a simple chain that summarizes a text. We will use this chain to explore the different Runnable methods.

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a concise summarization assistant."),
    ("human", "Summarize the following text in one sentence:\n\n{text}"),
])

chain = prompt | llm | StrOutputParser()
````

invoke: one input, one output

````python
chain.invoke({"text": "Hello world"})
````

batch: many inputs, processed in parallel

````python
texts = [
    {"text": "LangChain is a framework for building LLM apps."},
    {"text": "LangGraph lets you build stateful, graph-based agents."},
    {"text": "LangSmith helps you trace and debug your chains."},
]

summaries = chain.batch(texts, config={"max_concurrency": 3})

for s in summaries:
    print("-", s)
````

The `batch()` method allows you to process multiple inputs in a single call. By default, LangChain can execute these inputs concurrently, while the `max_concurrency` parameter lets you control the maximum number of simultaneous executions. This is particularly useful for managing performance and staying within your model provider's rate limits.

`stream()`: Display Output as It Is Generated

The `stream()` method allows you to receive the model's response progressively instead of waiting for the entire output to be generated.

This makes applications feel faster and more responsive because users can start reading the response as it is being generated, rather than waiting for the complete result.

````python
for chunk in chain.stream({"text": "Explain what LangChain is in three sentences."}):
    print(chunk, end="", flush=True)
````

ainvoke: async execution

In async applications (such as a FastAPI backend, covered in Article 7).

This unified interface is why LangChain components are so easy to connect, test, and run in parallel. It is also the same interface that LangGraph relies on.

## 13. Swapping Models Without Rewriting Your Code

Because all chat models follow the same `Runnable` interface, you can switch between different providers with minimal code changes.

LangChain provides a provider-agnostic helper called `init_chat_model`, which makes it easier to initialize and switch between language models.

````python
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini", model_provider="openai", temperature=0)

# Later, switch providers by changing one line (after installing the matching
# integration package, e.g. `pip install langchain-anthropic`):
# llm = init_chat_model("claude-sonnet-4-5", model_provider="anthropic", temperature=0)
````

The rest of your chain (prompt, parser, orchestration) remains untouched. This is the practical payoff of standard interfaces: your application logic does not depend on a single vendor.

## 14. Handling Errors and Invalid Inputs

LLM calls can fail for several reasons, including network issues, API rate limits, empty inputs, or unexpected responses. Proper error handling helps make your LangChain applications more reliable and robust.

### Basic Error Handling

Let's see how to use Python's `try` and `except` blocks to catch errors when executing a chain.

````python
try:
    result = chain.invoke({"text": ""})
    print(result)
except Exception as e:
    print("Error:", e)
````

### Built-in Retries

LangChain provides a built-in `with_retry()` method for Runnables. It automatically retries an operation when a failure occurs, helping make your chains more resilient to temporary errors such as network issues or API rate limits.

````python
reliable_chain = chain.with_retry(stop_after_attempt=3)

result = reliable_chain.invoke({"text": "Retry me if something goes wrong."})
````

### Fallbacks

If your primary model fails, fall back to another:

````python
primary = ChatOpenAI(model="gpt-4o-mini")
backup = ChatOpenAI(model="gpt-4o")

resilient_llm = primary.with_fallbacks([backup])

resilient_chain = prompt | resilient_llm | StrOutputParser()
````

Validate inputs before calling the model

Cheap checks save money and avoid weird outputs:

````python
def validate_text(inputs: dict) -> dict:
    text = inputs.get("text", "").strip()
    if len(text) < 10:
        raise ValueError("Please provide a text with at least 10 characters.")
    return {"text": text}

safe_chain = RunnableLambda(validate_text) | chain
````

Retries, fallbacks, and validation are small additions that make the difference between a demo and an application you can rely on.

## 16. Best Practices

Here are some best practices to keep in mind when building LangChain applications:

- **Keep prompts short and clear.** Clear instructions help improve the accuracy of model responses.
- **Reuse prompt templates.** Use one template with different inputs to simplify maintenance and debugging.
- **Chain small, testable steps.** Each component should have one responsibility, making the workflow easier to test and debug.
- **Test each component with `.invoke()` before chaining.** This helps identify issues early and makes debugging easier.
- **Use `temperature=0` for deterministic tasks.** Tasks such as extraction and classification benefit from consistent outputs.
- **Prefer structured output over string parsing.** Pydantic schemas provide more reliable and predictable results than regular expressions.
- **Pin your package versions.** Using `langchain>=1.0,<2.0` helps avoid unexpected breaking changes.
- **Never hard-code secrets.** Store API keys in `.env` files and exclude them from version control.
- **Enable tracing from day one.** Use LangSmith early to monitor executions and identify issues.

## 17. Mini Project: A Content Analyzer

Let's put everything together in a small, practical project.

We will build a content analyzer that takes an article as input and performs the following operations:

1. **Generate a summary** of the article.
2. **Extract keywords** to identify its main topics.
3. **Run both steps in parallel** to reduce execution time.
4. **Generate a catchy title** using the results of the previous steps.

````python
# 1. Import libraries and load environment variables
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnablePassthrough

load_dotenv()


# 2. Initialize the language model and output parser
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.3)
parser = StrOutputParser()


# 3. Define individual chains

# Chain 1: Generate a summary
summary_chain = (
    ChatPromptTemplate.from_template(
        "Summarize the following text in two sentences:\n\n{text}"
    )
    | llm
    | parser
)

# Chain 2: Extract keywords
keywords_chain = (
    ChatPromptTemplate.from_template(
        "Extract 5 relevant keywords from the following text. "
        "Return them as a comma-separated list:\n\n{text}"
    )
    | llm
    | parser
)

# Chain 3: Generate a title
title_chain = (
    ChatPromptTemplate.from_template(
        "Write a catchy title (max 10 words) for an article with this summary "
        "and these keywords.\n\nSummary: {summary}\nKeywords: {keywords}\n\n"
        "Return only the title."
    )
    | llm
    | parser
)


# 4. Build the complete pipeline

# Run summary and keyword extraction in parallel
# Then generate a title using both results
pipeline = (
    RunnableParallel(
        summary=summary_chain,
        keywords=keywords_chain
    )
    | RunnablePassthrough.assign(title=title_chain)
)


# 5. Provide the input article
article = """
LangChain is an open-source framework designed to simplify the development of
applications powered by large language models. It provides standard interfaces
for models, prompts, output parsers and tools, and lets developers compose them
into reusable pipelines. Built on top of it, LangGraph adds support for stateful,
graph-based workflows, which makes it possible to build agents that can loop,
branch, and collaborate.
"""


# 6. Execute the pipeline
result = pipeline.invoke({"text": article})


# 7. Display the results
print("TITLE   :", result["title"])
print("SUMMARY :", result["summary"])
print("KEYWORDS:", result["keywords"])
````

Look at how little code is needed and how clearly the data flows through the pipeline:

- **`RunnableParallel`** executes independent steps concurrently.
- **`RunnablePassthrough.assign()`** preserves the previous results and adds the generated title.
- **Each chain** is small, independently testable, and easy to replace.

Try extending the pipeline. Add a `sentiment` branch, convert the keywords into a Python list using `RunnableLambda`, or replace the output with a Pydantic model. This is how real-world applications evolve: by combining small, reusable components.

## Summary

- **LLM applications** combine models, logic, data, and tools. A model alone is not a complete application.
- **LangChain** provides the building blocks (models, prompts, parsers, and chains), while **LangGraph** manages the control flow between them.
- **Prompt templates** provide reusable prompt structures, and the **pipe operator (`|`)** connects components into chains using LCEL.
- Every component is a **Runnable**, sharing a common interface: `invoke`, `batch`, `stream`, and `ainvoke`.
- Use **structured output** instead of manual string parsing, **`RunnableParallel`** for independent tasks, and **`RunnableLambda`** for custom Python logic.
- **Retries, fallbacks, and validation** improve reliability, while **LangSmith** helps monitor and debug your applications.
